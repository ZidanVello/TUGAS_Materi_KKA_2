```python
import pandas as pd
import numpy as np
```

```python
harga = [5000,7000,3000, 12000, 4500]
harga_diskon = [h * 0.9 for h in harga]
print(harga_diskon)

harga = np.array([5000, 7000, 3000,12000,4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

**ANALISA 1**

Jika menggunakan list Python biasa, [5000,7000,3000] * 0.9 akan menghasilkan TypeError karena list tidak dapat dikalikan dengan bilangan desimal. Pada NumPy array, operasi * 0.9 diterapkan ke setiap elemen array



```python
data_kantin = {
'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
'harga': [12000, 4000, 10000, 4000, 8000],
'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
print(df)
```

**ANALISA 2**

kolom yang memiliki data kosong yaitu kolom menu dan terjual.resiko jika data langsung digunakan tanpa di bersihkan yaitu hasil nya pasti tidak akurat

```python
df = pd.read_csv("/content/data_kantin - data_kantin.csv")


print(df.head())
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape)
```

**ANALISA3**

kolom yang jumlah non-null yang lebih sedikit adalah menu dan terjual ,berati terdapat data yang kosong pada kolom menu dan terjual

```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu'])
```

**ANALISA 4**

untuk kolom terjual bisa diisi atau di set nilai 0 karena sama saja artinya dengan menu itu belum terjual.lalu untuk kolom menu itu adalah komponen penting atau wajib jadi tidak boleh diisi 0

```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int)
```

**ANALISA 5**

sebelum drop.duplicate nilai duplikat nya ada 5 lalu setelah di drop nilai nya 0. pentingnya memastikan tipe data adalah agar tidak terjadi error atau kesalahan misalnya kolom harga mungkin  tipe datanya berubah menjadi string kan bingung nanti nya

```python
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting
df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan
ringkasan = df.groupby('menu')['total_pendapatan'].sum()


print(ringkasan)
print('Maximum total pendapatan:', ringkasan.max())
```
